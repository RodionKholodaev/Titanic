# Baseline: предсказание только по полу

Женщина → выжила (1), мужчина → погиб (0).

```python
import pandas as pd

train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
def predict_by_sex(df):
    return (df['Sex'] == 'female').astype(int)

train_acc = (predict_by_sex(train) == train['Survived']).mean()
print(f'Accuracy на train: {train_acc:.4f}')
```

```python
submission = pd.DataFrame({
    'PassengerId': test['PassengerId'],
    'Survived': predict_by_sex(test),
})
submission.to_csv('baseline_submission.csv', index=False)
submission.head()
```
